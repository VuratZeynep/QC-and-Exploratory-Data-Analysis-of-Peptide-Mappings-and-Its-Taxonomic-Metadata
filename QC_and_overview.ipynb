# Load both Parquet Files

- Mapped peptide file contains the peptide mapping results. 
- Metadata file serves as a lookup table for annotations.

# QC Summary Table

```python
def QC(df,second_col):
    shape = df.shape

    missing_val = df.isna().sum()

    is_duplicated = df.duplicated().sum()

    return print(f"Number of rows and columns: {shape}\nMissing_values: {missing_val}\nDuplicates: {is_duplicated}\n")
    
```

```python
# mapped QC

print(QC(mapped, "position"))
```

```python
# metadata QC

print(QC(metadata, "taxid"))
```

# Column Overview

### 1. Mapped peptide

```python
mapped.columns
```

- peptide: Amino acid sequence.
- side: Represents whether the peptide belongs to a virus or a human.
- accesion: Unique identifiers for proteins assigned by the NCBI database.
- position: Starting position of the peptide within the protein.

### 2. Metadata

```python
metadata.columns
```

- accession: Unique identifiers for proteins assigned by the NCBI database.
- length: Length of the peptide chain. 
- taxid: Unique identifier assigned by the NCBI to the source organism of the protein. 
- title: Name of the protein or ORFs with the description of organism.
- species: Species name related with the protein.
- strain: Strain is a genetic variant, a subtype within a biological species.
- lineage_x: Taxonomic lineage of the species. 
- family: Family annotations of viruses. 
- lineage_y: Taxonomic lineage of the species.
